```python

# ============================================================
# 13 | ALTA10 WAKE-CONDITIONED ANEMOMETER EVIDENCE REVIEW
#
# ONE CELL — uses:
#   - Dataset 1: merged SCADA + wake classifications
#   - Notebook 08: farm-wide SCADA screening results
#
# Produces:
#   1. Wake-category turbine/peer comparison
#   2. Flagged-event wake and power cross-checks
#   3. Sponsor investigation review table
#   4. Wake-category coverage visualization
#
# Existing wake equations and 08 screening rules unchanged.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# ------------------------------------------------------------
# 1. SETTINGS
# ------------------------------------------------------------

ROOT_13 = Path(
    r"C:\Users\alex-\OneDrive\Documents\GitHub\CU-Capstone-Project"
)

MERGED_13 = (
    ROOT_13
    / "data/processed/12_farmwide_wake_exposure"
    / "scada_wake_comparison"
    / "ALTA10_SCADA_with_wake_categories.parquet"
)

BASE_08 = ROOT_13 / "data/processed/08_scada_screening"

OUTPUT_BASE_13 = (
    ROOT_13 / "data/processed/13_wake_conditioned_evidence"
)

MAST_MIN_13 = 7.0
MAST_MAX_13 = 10.0

SCADA_MIN_13 = 3.0
SCADA_MAX_13 = 25.0

# Original power measurements only — never KPI zero substitution.
POWER_MIN_13 = 100.0
POWER_MAX_13 = 2850.0

CATEGORIES_13 = ["LOW", "MODERATE", "HIGH"]

# ------------------------------------------------------------
# 2. FIND DATASET 1 AND PREVIOUS SCREENING RESULTS
# ------------------------------------------------------------

if not MERGED_13.is_file():
    raise FileNotFoundError(
        f"Run the Dataset 1 merge first:\n{MERGED_13}"
    )

runs_08 = sorted(
    [
        p for p in BASE_08.glob("run_*")
        if (
            (p / "08_turbine_review_table.csv").is_file()
            and (p / "08_flagged_episodes.csv").is_file()
            and (p / "08_fixed_peers.csv").is_file()
        )
    ],
    reverse=True
)

if not runs_08:
    raise FileNotFoundError(
        "No complete 08_scada_screening/run_* output found. "
        "Run the corrected Notebook 08 first."
    )

RUN_08 = runs_08[0]

review_08 = pd.read_csv(
    RUN_08 / "08_turbine_review_table.csv"
)

episodes_08 = pd.read_csv(
    RUN_08 / "08_flagged_episodes.csv"
)

peers_08 = pd.read_csv(
    RUN_08 / "08_fixed_peers.csv"
)

needed_review = {
    "turbine", "screening_label",
    "evaluation_peer_pairs", "evaluation_scored_records",
    "departure_days", "reasons"
}

needed_events = {
    "turbine", "reason",
    "first_timestamp_label_utc",
    "last_timestamp_label_utc",
    "records"
}

needed_peers = {"turbine", "peer", "rank"}

for name, frame, required in [
    ("08 review", review_08, needed_review),
    ("08 events", episodes_08, needed_events),
    ("08 peers", peers_08, needed_peers)
]:
    if not required.issubset(frame.columns):
        raise KeyError(
            f"{name} missing {sorted(required - set(frame.columns))}"
        )

# ------------------------------------------------------------
# 3. LOAD THE MERGED SCADA + WAKE DATASET
# ------------------------------------------------------------

cols_13 = [
    "timestamp",
    "turbine",
    "wind_speed",
    "real_power_ac",
    "mast_wind_speed_ms",
    "mast_wind_from_deg",
    "jensen_wake_category",
    "jensen_max_deficit_pct",
    "gaussian_style_wake_category"
]

d = pd.read_parquet(
    MERGED_13,
    columns=cols_13
)

d["timestamp"] = pd.to_datetime(
    d.timestamp,
    utc=True,
    errors="coerce"
)

if (
    d.timestamp.isna().any()
    or d.duplicated(["timestamp", "turbine"]).any()
):
    raise ValueError(
        "Dataset 1 has invalid or duplicate turbine timestamps."
    )

for col in [
    "wind_speed",
    "real_power_ac",
    "mast_wind_speed_ms",
    "mast_wind_from_deg",
    "jensen_max_deficit_pct"
]:
    d[col] = pd.to_numeric(
        d[col],
        errors="coerce"
    )

for col in [
    "jensen_wake_category",
    "gaussian_style_wake_category"
]:
    d[col] = (
        d[col]
        .astype("string")
        .str.strip()
        .str.upper()
    )

# Check that the dataset uses our agreed thresholds:
# Low <2%, Moderate 2–15%, High >15%.

finite_deficit = d.jensen_max_deficit_pct.notna()

expected_labels = np.where(
    d.jensen_max_deficit_pct.to_numpy() < 2,
    "LOW",
    np.where(
        d.jensen_max_deficit_pct.to_numpy() > 15,
        "HIGH",
        "MODERATE"
    )
)

if not np.array_equal(
    d.loc[
        finite_deficit,
        "jensen_wake_category"
    ].astype(str).to_numpy(),
    expected_labels[finite_deficit.to_numpy()]
):
    raise ValueError(
        "Dataset 1 classifications do not match "
        "LOW <2%, HIGH >15%. Rebuild the wake lookup/merge."
    )

# Match the original 08 investigation period.
d = d.loc[
    d.timestamp >= pd.Timestamp(
        "2026-05-01",
        tz="UTC"
    )
].copy()

if d.empty:
    raise ValueError(
        "No merged SCADA observations from May 2026 onward."
    )

print("Dataset 1 observations:", f"{len(d):,}")
print("Turbines:", d.turbine.nunique())
print("Previous screening run:", RUN_08)

# ------------------------------------------------------------
# 4. ORGANIZE MEASUREMENTS BY TIMESTAMP AND TURBINE
# ------------------------------------------------------------

wind = (
    d.pivot(
        index="timestamp",
        columns="turbine",
        values="wind_speed"
    )
    .sort_index()
)

power = d.pivot(
    index="timestamp",
    columns="turbine",
    values="real_power_ac"
).reindex(wind.index)

jwake = d.pivot(
    index="timestamp",
    columns="turbine",
    values="jensen_wake_category"
).reindex(wind.index)

gwake = d.pivot(
    index="timestamp",
    columns="turbine",
    values="gaussian_style_wake_category"
).reindex(wind.index)

mast = (
    d.drop_duplicates("timestamp")
    .set_index("timestamp")[
        ["mast_wind_speed_ms", "mast_wind_from_deg"]
    ]
    .reindex(wind.index)
)

# Confirm that the mast values are consistent across
# turbines sharing the same timestamp.

check_mast = d.groupby("timestamp")[
    ["mast_wind_speed_ms", "mast_wind_from_deg"]
].agg(["min", "max"])

for col in [
    "mast_wind_speed_ms",
    "mast_wind_from_deg"
]:
    a = check_mast[(col, "min")]
    b = check_mast[(col, "max")]

    if ((a - b).abs() > 1e-8).any():
        raise ValueError(
            f"Inconsistent {col} across turbines."
        )

available = set(wind.columns)

# Retrieve each turbine's original fixed three-peer group.
peer_sets = (
    peers_08.sort_values(["turbine", "rank"])
    .groupby("turbine")["peer"]
    .apply(list)
    .to_dict()
)

# ------------------------------------------------------------
# 5. COMPARE TURBINES WITH THEIR THREE FIXED PEERS
# ------------------------------------------------------------

# This comparison is descriptive across wake conditions.
# It is NOT a new sensor-fault classification.
#
# Only identical timestamps are compared.
# All three fixed peers must have valid wind-speed values.

valid_mast = mast.mast_wind_speed_ms.between(
    MAST_MIN_13,
    MAST_MAX_13
)

valid_wind = (
    wind.ge(SCADA_MIN_13)
    & wind.le(SCADA_MAX_13)
)

context_rows = []

for turbine in review_08.turbine:

    fixed = peer_sets.get(turbine, [])

    have_all = (
        turbine in available
        and len(fixed) == 3
        and all(p in available for p in fixed)
    )

    if have_all:

        simultaneous = (
            valid_mast
            & valid_wind[turbine]
            & valid_wind[fixed].all(axis=1)
        )

        peer_wind = wind[fixed].median(
            axis=1,
            skipna=False
        )

        raw_diff = wind[turbine] - peer_wind

        target_power = power[turbine]

        peer_all_low = (
            jwake[fixed].eq("LOW").all(axis=1)
        )

    else:

        simultaneous = pd.Series(
            False,
            index=wind.index
        )

        raw_diff = pd.Series(
            np.nan,
            index=wind.index
        )

        target_power = pd.Series(
            np.nan,
            index=wind.index
        )

        peer_all_low = pd.Series(
            False,
            index=wind.index
        )

    for category in CATEGORIES_13:

        if turbine in available:
            mask = (
                simultaneous
                & jwake[turbine].eq(category)
            )
        else:
            mask = simultaneous

        values = raw_diff.loc[mask]

        n = int(mask.sum())

        context_rows.append({
            "turbine": turbine,
            "jensen_wake_category": category,

            "simultaneous_wind_n": n,

            "distinct_days": int(
                values.index.normalize().nunique()
            ),

            "mean_target_minus_3peer_ms":
                values.mean(),

            "median_target_minus_3peer_ms":
                values.median(),

            "mean_mast_speed_ms":
                mast.loc[
                    mask,
                    "mast_wind_speed_ms"
                ].mean(),

            "original_power_100to2850_n": int(
                target_power.loc[mask]
                .between(
                    POWER_MIN_13,
                    POWER_MAX_13
                )
                .sum()
            ),

            "all_3_peers_predicted_low_pct":
                (
                    100 * peer_all_low.loc[mask].mean()
                    if n else np.nan
                ),

            "fixed_peers": "|".join(fixed)
        })

wake_context = pd.DataFrame(context_rows)

# ------------------------------------------------------------
# 6. RECHECK THE PREVIOUSLY FLAGGED EPISODES
# ------------------------------------------------------------

# Reuse events from 08 instead of inventing new flags.
#
# Check:
# - Jensen / Gaussian wake conditions
# - Target wind versus fixed peer median
# - Original measured power versus fixed peer median
#
# No KPI zero substitutions.

event_rows = []

for row in episodes_08.itertuples(index=False):

    turbine = str(row.turbine)
    fixed = peer_sets.get(turbine, [])

    start = pd.to_datetime(
        row.first_timestamp_label_utc,
        utc=True
    )

    end = pd.to_datetime(
        row.last_timestamp_label_utc,
        utc=True
    )

    times = pd.date_range(
        start,
        end,
        freq="10min",
        tz="UTC"
    )

    if len(times) != int(row.records):
        raise ValueError(
            f"Event duration/record mismatch: {turbine}, {start}"
        )

    if (
        turbine not in available
        or len(fixed) != 3
        or any(p not in available for p in fixed)
    ):
        continue

    # Target and peer wind speeds
    target_speed = wind[turbine].reindex(times)
    peer_speeds = wind[fixed].reindex(times)

    peer_speed_median = peer_speeds.median(
        axis=1,
        skipna=False
    )

    peer_speed_valid = (
        peer_speeds.ge(SCADA_MIN_13)
        & peer_speeds.le(SCADA_MAX_13)
    ).all(axis=1)

    speed_ok = (
        target_speed.between(
            SCADA_MIN_13,
            SCADA_MAX_13
        )
        & peer_speed_valid
    )

    speed_diff = (
        target_speed - peer_speed_median
    ).where(speed_ok)

    # Original electrical power
    target_power = power[turbine].reindex(times)
    peer_powers = power[fixed].reindex(times)

    peer_power_valid = (
        peer_powers.ge(POWER_MIN_13)
        & peer_powers.le(POWER_MAX_13)
    ).all(axis=1)

    power_ok = (
        target_power.between(
            POWER_MIN_13,
            POWER_MAX_13
        )
        & peer_power_valid
    )

    power_diff = (
        target_power
        - peer_powers.median(
            axis=1,
            skipna=False
        )
    ).where(power_ok)

    both = (
        speed_diff.notna()
        & power_diff.notna()
    )

    sign_agreement = (
        (
            (speed_diff[both] > 0)
            & (power_diff[both] > 0)
        )
        |
        (
            (speed_diff[both] < 0)
            & (power_diff[both] < 0)
        )
    )

    # Joint low-wake status for target + all three peers
    jensen_joint_low = (
        jwake[turbine].reindex(times).eq("LOW")
        & jwake[fixed].reindex(times).eq("LOW").all(axis=1)
    )

    gaussian_joint_low = (
        gwake[turbine].reindex(times).eq("LOW")
        & gwake[fixed].reindex(times).eq("LOW").all(axis=1)
    )

    event_rows.append({
        "turbine": turbine,
        "08_reason": row.reason,
        "start_utc": start,
        "last_timestamp_utc": end,

        "nominal_minutes": 10 * len(times),
        "records": len(times),

        "jensen_joint_all_four_low_pct":
            100 * jensen_joint_low.mean(),

        "gaussian_joint_all_four_low_pct":
            100 * gaussian_joint_low.mean(),

        "wind_comparable_n":
            int(speed_diff.notna().sum()),

        "mean_target_minus_peers_wind_ms":
            speed_diff.mean(),

        "original_power_comparable_n":
            int(power_diff.notna().sum()),

        "mean_target_minus_peers_power_kw":
            power_diff.mean(),

        "both_wind_and_power_n":
            int(both.sum()),

        "matching_wind_power_difference_sign_pct":
            (
                100 * sign_agreement.mean()
                if both.any()
                else np.nan
            ),

        "peers": "|".join(fixed)
    })

EVENT_COLUMNS_13 = [
    "turbine",
    "08_reason",
    "start_utc",
    "last_timestamp_utc",
    "nominal_minutes",
    "records",
    "jensen_joint_all_four_low_pct",
    "gaussian_joint_all_four_low_pct",
    "wind_comparable_n",
    "mean_target_minus_peers_wind_ms",
    "original_power_comparable_n",
    "mean_target_minus_peers_power_kw",
    "both_wind_and_power_n",
    "matching_wind_power_difference_sign_pct",
    "peers"
]

event_check = pd.DataFrame(
    event_rows,
    columns=EVENT_COLUMNS_13
)

# ------------------------------------------------------------
# 7. BUILD THE SPONSOR INVESTIGATION TABLE
# ------------------------------------------------------------

counts = (
    wake_context
    .pivot(
        index="turbine",
        columns="jensen_wake_category",
        values="simultaneous_wind_n"
    )
    .reindex(columns=CATEGORIES_13)
    .fillna(0)
    .astype(int)
    .rename(
        columns={
            c: f"{c.lower()}_wake_peer_records_7to10"
            for c in CATEGORIES_13
        }
    )
)

if event_check.empty:

    event_agg = pd.DataFrame(
        columns=[
            "turbine",
            "crosschecked_08_episodes",
            "episodes_with_power_context",
            "mean_event_power_sign_agreement_pct"
        ]
    )

else:

    event_agg = (
        event_check.groupby("turbine")
        .agg(
            crosschecked_08_episodes=(
                "08_reason",
                "size"
            ),

            episodes_with_power_context=(
                "original_power_comparable_n",
                lambda x: int(x.gt(0).sum())
            ),

            mean_event_power_sign_agreement_pct=(
                "matching_wind_power_difference_sign_pct",
                "mean"
            )
        )
        .reset_index()
    )

review = (
    review_08
    .merge(
        counts.reset_index(),
        on="turbine",
        how="left",
        validate="one_to_one"
    )
    .merge(
        event_agg,
        on="turbine",
        how="left",
        validate="one_to_one"
    )
)

for col in [
    "crosschecked_08_episodes",
    "episodes_with_power_context"
]:
    review[col] = (
        review[col].fillna(0).astype(int)
    )

# Preserve 08 classifications and investigation order.
order = {
    "Priority investigation": 0,
    "Monitor / context required": 1,
    "Insufficient evidence": 2,
    "No strong anomaly in assessed conditions": 3
}

review["_order"] = (
    review.screening_label
    .map(order)
    .fillna(4)
)

review = (
    review.sort_values(
        ["_order", "departure_days", "turbine"],
        ascending=[True, False, True]
    )
    .drop(columns="_order")
)

# ------------------------------------------------------------
# 8. SAVE ALL RESULTS
# ------------------------------------------------------------

OUT_13 = OUTPUT_BASE_13 / (
    "run_"
    + datetime.now(timezone.utc).strftime(
        "%Y%m%dT%H%M%S%fZ"
    )
)

OUT_13.mkdir(
    parents=True,
    exist_ok=False
)

wake_context.to_csv(
    OUT_13 / "13_wake_category_peer_comparison.csv",
    index=False
)

event_check.to_csv(
    OUT_13 / "13_flagged_event_power_wake_checks.csv",
    index=False
)

review.to_csv(
    OUT_13 / "13_sponsor_investigation_table.csv",
    index=False
)

# ------------------------------------------------------------
# 9. WAKE-CATEGORY COVERAGE GRAPH
# ------------------------------------------------------------

chart = review[
    ["turbine", "screening_label"]
].copy()

for category in CATEGORIES_13:

    chart[category] = review[
        f"{category.lower()}_wake_peer_records_7to10"
    ].fillna(0).to_numpy()

totals = (
    chart[CATEGORIES_13]
    .sum(axis=1)
    .replace(0, np.nan)
)

fig, ax = plt.subplots(
    figsize=(
        11,
        max(9, len(chart) * 0.26)
    )
)

left = np.zeros(len(chart))

for category, color in [
    ("LOW", "#2ca25f"),
    ("MODERATE", "#fdae61"),
    ("HIGH", "#d73027")
]:

    pct = (
        100 * chart[category] / totals
    ).fillna(0).to_numpy()

    ax.barh(
        np.arange(len(chart)),
        pct,
        left=left,
        color=color,
        label=category
    )

    left += pct

ax.set_yticks(
    np.arange(len(chart)),
    chart.turbine,
    fontsize=8
)

ax.invert_yaxis()
ax.set_xlim(0, 100)

ax.set_xlabel(
    "Share of simultaneous three-peer wind observations, "
    "mast 7–10 m/s (%)"
)

ax.set_title(
    "ALTA10 — Wake-Category Coverage by Turbine\n"
    "Existing 08 investigation order"
)

ax.legend(loc="lower right")
ax.grid(axis="x", alpha=0.15)

fig.tight_layout()

fig.savefig(
    OUT_13 / "13_wake_category_peer_coverage.png",
    dpi=180,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

# ------------------------------------------------------------
# 10. SAVE METHODOLOGY AND INTERPRETATION NOTES
# ------------------------------------------------------------

notes = f"""
13 | ALTA10 WAKE-CONDITIONED EVIDENCE REVIEW

Source screening run:
{RUN_08}

Source merged SCADA/wake dataset:
{MERGED_13}

This notebook does not rerun or change the 08 screening
thresholds or early/later baseline. Investigation labels
are copied directly from 08.

WAKE COMPARISON
----------------
Compare each turbine with its three fixed peers at
identical timestamps when mast speed is 7–10 m/s and
reported SCADA wind speed is 3–25 m/s for all four.

LOW: predicted deficit <2%
MODERATE: predicted deficit 2–15%
HIGH: predicted deficit >15%

The categories use the existing Jensen maximum across
the five wind-direction sensitivity offsets.

Comparisons across wake categories are descriptive.
Different wind directions, terrain, and operating states
may affect observed wind-speed differences.

A turbine's moderate/high category does not require
all three peers to share that category.

EXISTING 08 FLAGS
------------------
The original 08 detector only assessed jointly low-wake
periods with additional power-range eligibility.

Therefore previously flagged events are expected to be
low wake by design. This does NOT demonstrate that
anomalies are more common in low-wake conditions.

POWER CROSS-CHECK
------------------
Only original real_power_ac measurements are used.
Missing original power is never replaced by KPI zeros.

Power differences provide context, not proof of sensor
accuracy or instrument malfunction.

LIMITATIONS
------------
The mast and turbines are at different locations.
Fixed nearby peers are not calibrated reference sensors.
The wake model is simplified and uncalibrated.
An earlier baseline may contain persistent unknown bias.
No strong anomaly means only that the implemented
rules did not trigger under assessed conditions.

WTG159's LiDAR anomaly is a separate measurement issue.
It need not be detected in this SCADA-only comparison.
"""

(
    OUT_13 / "13_README_interpretation.txt"
).write_text(
    notes,
    encoding="utf-8"
)

# ------------------------------------------------------------
# 11. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("13 | WAKE-CONDITIONED EVIDENCE REVIEW COMPLETE")
print("=" * 72)

print("\nOriginal screening results:")
print(RUN_08)

print("\nTurbines reviewed:", len(review))
print("Original 08 events:", len(episodes_08))
print("Cross-checked events:", len(event_check))

print("\nInvestigation labels (UNCHANGED from 08):")

print(
    review.screening_label
    .value_counts()
    .to_string()
)

viewcols = [
    "turbine",
    "screening_label",
    "evaluation_peer_pairs",
    "evaluation_scored_records",
    "departure_days",
    "low_wake_peer_records_7to10",
    "moderate_wake_peer_records_7to10",
    "high_wake_peer_records_7to10",
    "crosschecked_08_episodes",
    "episodes_with_power_context",
    "reasons"
]

print("\nWTG159 AND WTG172 — INVESTIGATION REVIEW")

display(
    review.loc[
        review.turbine.isin(
            ["WTG159", "WTG172"]
        ),
        viewcols
    ].round(2)
)

print("\nTOP 12 TURBINES — EXISTING 08 PRIORITY ORDER")

display(
    review[viewcols].head(12).round(2)
)

print("\nFIRST FLAGGED EVENT CROSS-CHECKS")

display(
    event_check.head(12).round(2)
)

print("\nALL RESULTS SAVED TO:")
print(OUT_13)
```
