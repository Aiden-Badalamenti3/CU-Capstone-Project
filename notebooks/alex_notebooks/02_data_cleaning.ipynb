```python
import pandas as pd
import re

# --------------------------------------------------
# 1. CLEAN COLUMN NAMES
# --------------------------------------------------

def clean_column(col):
    # Special case for time column
    if col == "utcTime":
        return "utc_time"

    # Remove site prefix
    col = col.replace("ALTA10-SCEWF-", "")

    # Remove KPI marker
    col = col.replace("-KPI-", "-")

    # WTG169 -> wtg169 without splitting W_T_G
    col = re.sub(r'WTG(\d+)', r'wtg\1', col)

    # Replace hyphens with underscores
    col = col.replace("-", "_")

    # Convert CamelCase to snake_case
    col = re.sub(r'(?<=[a-z0-9])(?=[A-Z])', '_', col)

    # Add underscore before trailing position numbers
    col = re.sub(r'([A-Za-z])(\d+)$', r'\1_\2', col)

    # Clean duplicates and lowercase
    col = re.sub(r'_+', '_', col).lower().strip('_')

    return col


# Make a copy so df1 remains untouched
df_clean = df1.copy()

# Clean the column names
df_clean.columns = [clean_column(col) for col in df_clean.columns]


# --------------------------------------------------
# 2. CALCULATE MISSING DATA
# --------------------------------------------------

missing_summary = pd.DataFrame({
    "missing_count": df_clean.isna().sum(),
    "missing_percent": df_clean.isna().mean() * 100
}).sort_values("missing_percent", ascending=False)


# --------------------------------------------------
# 3. CHOOSE A THRESHOLD
# --------------------------------------------------

threshold = 80   # hide columns with more than 80% missing


# Identify them, but DO NOT delete them from df_clean
high_missing_cols = missing_summary[
    missing_summary["missing_percent"] > threshold
].index.tolist()


# --------------------------------------------------
# 4. CREATE A WORKING VIEW
# --------------------------------------------------

# This DataFrame hides the high-missing columns
df_working = df_clean.drop(columns=high_missing_cols)


# --------------------------------------------------
# 5. SUMMARY
# --------------------------------------------------

print("Original df1 shape:", df1.shape)
print("Cleaned df shape:", df_clean.shape)
print("Working df shape:", df_working.shape)

print(
    f"\nColumns hidden because they are > {threshold}% missing:",
    len(high_missing_cols)
)

print("\nTop missing columns:")
display(missing_summary.head(20))
```

```python
import pandas as pd
import re

rows = []

for i, col in enumerate(df_clean.columns):

    if col == "utc_time":
        continue

    # Works with names like w_t_g169_real_power_a_c
    match = re.match(r"(w_t_g\d+)_(.+)", str(col))

    if match:
        turbine = match.group(1)
        variable = match.group(2)

        # Access by column POSITION so duplicate names aren't a problem
        column_data = df_clean.iloc[:, i]

        rows.append({
            "turbine": turbine,
            "variable": variable,
            "column_name": col,
            "column_position": i,
            "missing_count": column_data.isna().sum(),
            "missing_percent": column_data.isna().mean() * 100
        })

missing_by_turbine = pd.DataFrame(rows)

missing_by_turbine = (
    missing_by_turbine
    .sort_values(
        ["turbine", "missing_percent"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

missing_by_turbine
```

```python
import pandas as pd
from collections import Counter

# --------------------------------------------------
# MAKE DUPLICATE COLUMN NAMES UNIQUE
# --------------------------------------------------

counts = Counter()
unique_columns = []

for col in df_clean.columns:
    counts[col] += 1

    if counts[col] == 1:
        unique_columns.append(col)
    else:
        unique_columns.append(f"{col}_{counts[col]}")

df_clean.columns = unique_columns


# --------------------------------------------------
# CHECK FOR DUPLICATES
# --------------------------------------------------

print(
    "Duplicate column names:",
    df_clean.columns.duplicated().sum()
)

# Optional: see the final column names
for col in df_clean.columns:
    print(col)
```

```python
import pandas as pd
import re

# Load processed data
df = pd.read_parquet(
    "../../data/processed/alta10_turbine_data_cleaned.parquet"
)

# Use time as the index
df_turbines = df.set_index("utc_time")

new_columns = []

for col in df_turbines.columns:

    # Match columns like w_t_g150_wind_speed
    match = re.match(r"(w_t_g\d+)_(.+)", col)

    if match:
        turbine = match.group(1)
        variable = match.group(2)

        # Rename w_t_g150 -> wtg150
        turbine = turbine.replace("w_t_g", "wtg")

        new_columns.append((turbine, variable))

    else:
        new_columns.append(("other", col))

# Create grouped turbine / variable columns
df_turbines.columns = pd.MultiIndex.from_tuples(
    new_columns,
    names=["turbine", "variable"]
)

# Sort turbines together
df_turbines = df_turbines.sort_index(axis=1)

print(df_turbines.shape)

df_turbines.head()
```

```python
import os

# Create processed folder if needed
processed_dir = "../../data/processed"
os.makedirs(processed_dir, exist_ok=True)

# Save the grouped turbine dataframe
output_path = os.path.join(
    processed_dir,
    "alta10_turbine_data_grouped.parquet"
)

df_turbines.to_parquet(output_path)

print("Saved to:")
print(output_path)

print("\nShape:")
print(df_turbines.shape)
```
