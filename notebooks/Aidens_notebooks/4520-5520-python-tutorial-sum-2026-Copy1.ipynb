### **(a) Numpy**

NumPy has become one of the foundational layers of the entire scientific Python ecosystem. Even when you aren't using it directly, there's a good chance it's doing work behind the scenes.

A typical dependency stack looks something like this:

```txt
Python
    │
 NumPy  ← n-dimensional arrays, broadcasting, vectorized operations
    │
 ├── SciPy          (optimization, FFT, linear algebra, statistics)
 ├── scikit-image   (image processing)
 ├── OpenCV         (can interoperate directly with NumPy arrays)
 ├── pandas         (built on NumPy arrays)
 ├── scikit-learn   (machine learning)
 ├── matplotlib     (plotting)
 └── PyTorch        (easy conversion to/from NumPy arrays on CPU)
```

import yfinance as yf
import matplotlib.pyplot as plt

# Define stock tickers and temporal range
tickers = ["TSLA", "GOOGL"]
start_date = "2025-01-01"
end_date = "2026-06-25"

# Fetch historical data
data = yf.download(tickers, start=start_date, end=end_date)

# Extract adjusted closing prices
close_prices = data['Close']

# Normalize prices to start at 100 for a direct performance comparison
normalized_prices = (close_prices / close_prices.iloc[0]) * 100

# Generate the visualization plot
plt.figure(figsize=(12, 6))
plt.plot(normalized_prices.index, normalized_prices['TSLA'], label='Tesla (TSLA)', color='#E82127', linewidth=2)
plt.plot(normalized_prices.index, normalized_prices['GOOGL'], label='Google (GOOGL)', color='#4285F4', linewidth=2)

# Format the chart aesthetics
plt.title('Tesla vs. Google Stock Performance Comparison (Base 100)', fontsize=14, fontweight='bold')
plt.xlabel('Timeline Date', fontsize=12)
plt.ylabel('Normalized Price (Base 100)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper left', fontsize=11)

# Display the final time series graph
plt.tight_layout()
plt.show()


````python
# 3. Alternative Raw Value Plotting
# Plots absolute dollar values instead of comparative percentages
fig, ax = plt.subplots(figsize=(10,5))
ax.plot(close_prices.index, close_prices['TSLA'],  'r-', label='TSLA ($)')
ax.plot(close_prices.index, close_prices['GOOGL'], 'b-', label='GOOGL ($)')
ax.set_title('Sample Visualization: TSLA vs GOOGL TS (base 100)',c='b',fontweight='bold')
ax.legend()
ax.set(xlabel='Date Timeline',ylabel='Normalized Price')
ax.grid(ls='--')
plt.tight_layout(); plt.show()
````

## **Sklearn Datasets**

````python
from sklearn import datasets

iris = datasets.load_iris()

import matplotlib.pyplot as plt

_, ax = plt.subplots()
scatter = ax.scatter(iris.data[:, 0], iris.data[:, 1], c=iris.target)
ax.set(xlabel=iris.feature_names[0], ylabel=iris.feature_names[1])
_ = ax.legend(
    scatter.legend_elements()[0], iris.target_names, loc="lower right", title="Classes"
)
plt.show()
````
