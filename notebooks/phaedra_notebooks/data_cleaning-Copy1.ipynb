```python
# defining a function to mark missing date/times in the datasets
def missing_times(df, year, month):
    # defines the starting time of 10 minute incriments
    start = pd.Timestamp(year = year, month = month, day = 1)
    end = start + pd.offsets.MonthBegin(1) - pd.Timedelta(minutes=10)
    expected = pd.date_range(start, end, freq = '10min')

    # calculates missing values
    missing = expected.difference(df.index)

    # returns missing datetimes
    return missing
```

### Converting Turbine Datasets to Long Format and Changing Variable Names

```python
# reordering and renaming variables with a dictionary
mapping1 = {'utcTime': 'utc', 'turbine': 'turbine', 'AirDensity': 'air_density', 'KPI-TemperatureAmbient-Merged': 'KPI_temp_ambient_merged', 
           'KPI-TemperatureAmbient': 'KPI_temp_ambient', 'KPI-TheoreticalPower': 'KPI_theoretical_power', 'KPI-RealPowerAC': 'KPI_real_power_ac',
           'RealPowerAC': 'real_power_ac', 'KPI-WindSpeed': 'KPI_wind_speed', 'WindSpeed': 'wind_speed', 'WindDirection': 'wind_direction', 
           'WindDirectionRelative': 'wind_direction_relative', 'NacellePosition': 'nacelle_pos', 'BladePosition1': 'blade_1_pos', 
           'BladePosition2': 'blade_2_pos', 'BladePosition3': 'blade_3_pos', 'BladeSetpointPosition1': 'blade_1_setpos', 
           'BladeSetpointPosition2': 'blade_2_setpos', 'BladeSetpointPosition3': 'blade_3_setpos'}

# defining a function to convert the wide format datasets into long format datasets
def reshape_month_turbine(df):
    # defines utcTime as index for rows in new dataset
    df = df.set_index('utcTime')

    # splits the columns into the turbine name and variable name
    parts = df.columns.str.extract(r"^ALTA\d+-SCEWF-(WTG\d+)-(.+)$")
    parts.columns = ['turbine', 'variable']

    # outputs problematic columns from this split
    bad = df.columns[parts['turbine'].isna()]
    if len(bad):
        raise ValueError(f"Columns that didn't match the pattern: {list(bad)}")

    # creates a copy of the dataframe
    df = df.copy()
    df.columns = pd.MultiIndex.from_frame(parts)

    # creates the long version of the dataset
    long = df.stack(level = 'turbine', future_stack = True).reset_index()
    long['turbine'] = long['turbine'].astype(str)
    long.columns.name = None
    long.sort_values(['turbine', 'utcTime']).reset_index(drop = True)

    # returns new dataset with renamed and reordered columns
    return long[list(mapping1)].rename(columns = mapping1)
```

```python
# applying the reshape_month_turbine() function to each monthly dataset
jan = reshape_month_turbine(jan_wide)
feb = reshape_month_turbine(feb_wide)
mar = reshape_month_turbine(mar_wide)
apr = reshape_month_turbine(apr_wide)
may = reshape_month_turbine(may_wide)
jun = reshape_month_turbine(jun_wide)
jul = reshape_month_turbine(jul_wide)
aug = reshape_month_turbine(aug_wide)
sep = reshape_month_turbine(sep_wide)
```

### Cleaning LIDAR Datasets

```python
# reordering and renaming variables with a dictionary
mapping4 = {'utc': 'utc', 'turbine': 'turbine', 'v': 'wind_speed_l', 'v_std': 'wind_speed_l_std', 'phi': 'wind_direction_l', 
            'availability': 'n_available_l', 'control': 'control', 'sensor1_v': 'wind_speed_l1', 'sensor1_v_std': 'wind_speed_l1_std', 
            'sensor2_v': 'wind_speed_l2', 'sensor2_v_std': 'wind_speed_l2_std', 'sensor1_phi': 'wind_direction_l1', 
            'sensor2_phi': 'wind_direction_l2'}

# creates a function for renaming the columns and changing dtype of Timestamp
def datetime_set(df, turbine):
    df = df

    # turn the Timestamp variable into a datetime object
    df['Timestamp'] = pd.to_datetime(df['Timestamp'])
    df['utc'] = df['Timestamp'] +  pd.Timedelta(hours = 8)
    df['turbine'] = turbine
    
    # returns new dataset with renamed and reordered columns
    return df[list(mapping4)].rename(columns = mapping4)
```

```python
# creating a function to highlight overlapping observations
def overlap(df1, df2):
    # inner merge keeps only rows whose keys appear in both DataFrames
    merged = df1.merge(df2, on = 'utc', suffixes = ('_1', '_2'))
    cols = [c for c in df1.columns if c != 'utc' and c in df2.columns]

    diffs = merged[['utc']].copy()
    mismatch = pd.Series(False, index = merged.index)

    for col in cols:
        a, b = merged[f'{col}_1'], merged[f'{col}_2']
        both_nan = a.isna() & b.isna()   # treat NaN vs NaN as a match

        if pd.api.types.is_numeric_dtype(a) and pd.api.types.is_numeric_dtype(b):
            diffs[col] = a - b
            differs = ~((a - b).abs() <= 1e-9) & ~both_nan
        else:
            differs = a.ne(b) & ~both_nan
            diffs[col] = differs

        mismatch |= differs

    print(f'{len(merged)} overlapping rows, {mismatch.sum()} with differences')
    return diffs[mismatch]
```

```python
# creating a function that combines the datasets while removing overlapped observations
def combine(dfs):
    combined = pd.concat(dfs, ignore_index = True)
    combined = combined.drop_duplicates(subset = 'utc', keep = 'first')
    return combined.sort_values('utc').reset_index(drop = True)
```

```python
def missing_intervals(df_full, turbine, var):
    # defining dataset
    df = df_full[df_full['turbine'] == turbine]
    
    # define stepcount for dataset and sort by datetime
    step = pd.Timedelta(minutes = 10)
    df = df.sort_values('utc').reset_index(drop = True)

    # define missing values
    is_missing = df[var].isna()
    run_id = (is_missing != is_missing.shift()).cumsum()

    # summarize values from before the outages
    roll = df.set_index('utc')[var].rolling('30min', closed = 'left')
    df['prev_30_avg'] = roll.mean().to_numpy()
    df['prev_30_n'] = roll.count().to_numpy()
    df['prev_value'] = df[var].shift(1)

    # summarizing each interval
    summary = (df[is_missing].groupby(run_id[is_missing]).agg(start = ('utc', 'first'), obs_count = ('utc', 'size'), 
                                                              prev_value = ('prev_value', 'first'), prev_30_avg = ('prev_30_avg', 'first'), 
                                                              prev_30_n = ('prev_30_n', 'first')).reset_index(drop = True))
    summary['length'] = summary['obs_count'] * step
    summary['end'] = summary['start'] + summary['length']

    return summary
```

```python
result = missing_intervals(scada, 'WTG170', 'real_power_ac')

result
```

```python
def top_gaps(summary, n=10):
    top = (
        summary.sort_values('length', ascending=False)
               .head(n)
               .reset_index(drop=True)
    )
    top.index = top.index + 1   # rank starting at 1 instead of 0

    table = pd.DataFrame({
        'Start': top['start'].dt.strftime('%Y-%m-%d %H:%M'),
        'End': top['end'].dt.strftime('%Y-%m-%d %H:%M'),
        'Duration (hours)': (top['length'].dt.total_seconds() / 3600).round(1),
        'Missing obs': top['obs_count'],
        'Last value before gap': top['prev_value'].round(1),
        'Avg 30 min before': top['prev_30_avg'].round(1),
    })
    table.index.name = 'Rank'
    return table
```
