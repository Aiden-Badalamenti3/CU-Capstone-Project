## Importing Libraries

```python
# load in turbine and met masts datasets
scada = pd.read_parquet('../../data/cleaned/scada_data.parquet')
scada_noloss = pd.read_parquet('../../data/cleaned/scada_noloss.parquet')
metmasts = pd.read_parquet('../../data/cleaned/metmasts_data.parquet')
reference = pd.read_parquet('../../data/cleaned/reference_power_curve.parquet')
wake = pd.read_parquet('../../data/cleaned/ALTA10_SCADA_with_wake_categories.parquet')
```

## Defining Functions

```python
def add_metmast(data, metmast):
    cols = ['turbine', 'utc', 'wind_speed']

    a = data[cols].assign(utc = lambda d: pd.to_datetime(d['utc'], utc = True))
    b = metmast[cols].assign(utc = lambda d: pd.to_datetime(d['utc'], utc = True))

    combined = pd.concat([a, b], ignore_index = True).sort_values('utc', ignore_index = True)

    group_ranges = {1: (150, 159), 2: (161, 169), 3: (170, 177),
                4: (178, 181), 5: (182, 192), 6: (193, 197)}
    rows = [(f"WTG{n}", g) for g, (lo, hi) in group_ranges.items() for n in range(lo, hi + 1)]
    rows.append(('tower', 1))
    grouping = pd.DataFrame(rows, columns = ['turbine', 'group'])

    combined_all = combined.merge(grouping, on = 'turbine', how = 'left').sort_values('utc', ignore_index = True)

    return combined_all
```

```python
def power_ratios(scada, ref, power_col = 'KPI_real_power_ac', ws_min = 3, ws_max = 26):
    def snap(values, levels):
        arr = np.asarray(values, dtype = float)
        idx = np.abs(arr[:, None] - levels).argmin(axis = -1)
        return np.where(np.isnan(arr), np.nan, levels[idx])

    df = scada.loc[scada['wind_speed'].between(ws_min, ws_max)].copy()

    for col in ['air_density', 'wind_speed']:
        levels = np.sort(ref[f'{col}_ref'].dropna().unique())
        df[f'{col}_ref'] = snap(df[col], levels)

    out = (df.merge(reference, on = ['wind_speed_ref', 'air_density_ref'], how = 'left').sort_values('utc', ignore_index = True))

    out['power_ratio_general'] = out[power_col] / out['power_ref'].replace(0, np.nan)
    out['power_ratio_theoretical'] = (out[power_col] / out['KPI_theoretical_power'].replace(0, np.nan))
    
    return out
```

```python
def corrected_wind_speed(wind_speed, air_density, ws0 = 7, ws1 = 12, ex0 = 1/3, ex1 = 2/3, standard = 1.050817):
    frac = np.clip((wind_speed - ws0) / (ws1 - ws0), 0, 1)
    exponent = ex0 + frac * (ex1 - ex0)
    
    return wind_speed * (air_density / standard) ** exponent
```

```python
def avg_power_curve(df_full, turbine, p = 'M', cutin = 4, rated = 15, width = 0.5, wind_speed = 'corrected_ws', min_count = 1, 
                    exclude_target_from_fleet = False):
    # filter to the given wind-speed range.
    valid_ws = (df_full[wind_speed].ge(cutin) & df_full[wind_speed].le(rated))
    df_all = df_full.loc[valid_ws].copy()

    # create a common period and bin definition.
    df_all['period'] = df_all['utc'].dt.to_period(p)
    n_bins = int(np.ceil((rated - cutin) / width))
    df_all['bin'] = np.floor((df_all[wind_speed] - cutin) / width).astype('int64').clip(0, n_bins - 1)

    # target turbine's period-by-bin averages.
    df = df_all.loc[df_all['turbine'].eq(turbine)].copy()

    bin_means = (df.groupby(['period', 'bin'], observed = True).agg(mean_real_power = ('KPI_real_power_ac', 'mean'),
                                                                    mean_theoretical_power = ('KPI_theoretical_power', 'mean'),
                                                                    n = ('KPI_real_power_ac', 'count')).reset_index())

    # fleet reference: mean measured power for each period and bin.
    fleet_data = df_all

    if exclude_target_from_fleet:
        fleet_data = fleet_data.loc[~fleet_data['turbine'].eq(turbine)]

    fleet_means = (fleet_data.groupby(['period', 'bin'], observed = True).agg(mean_fleet_power = ('KPI_real_power_ac', 'mean'),
                                                                              n_fleet = ('KPI_real_power_ac', 'count'),
                                                                              n_fleet_turbines = ('turbine', 'nunique')).reset_index())

    # join the reference values by their matching period and bin.
    bin_means = bin_means.merge(fleet_means, on = ['period', 'bin'], how = 'left', validate = 'one_to_one')

    # wind-speed bin midpoint.
    bin_means[wind_speed] = (cutin + (bin_means['bin'] + 0.5) * width)

    # correct the midpoint for a potentially partial final bin.
    bin_means[wind_speed] = np.minimum(bin_means[wind_speed], rated)

    bin_means['turbine'] = turbine

    # calculate relative differences; invalid references become NaN.
    valid_target = (bin_means['n'].ge(min_count) & bin_means['mean_real_power'].notna())
    valid_theoretical = (valid_target & bin_means['mean_theoretical_power'].gt(0))
    valid_fleet = (valid_target & bin_means['mean_fleet_power'].gt(0))

    bin_means['perc_diff_theoretical'] = np.nan
    bin_means.loc[valid_theoretical, 'perc_diff_theoretical'] = (bin_means.loc[valid_theoretical, 'mean_real_power']
                                                                 / bin_means.loc[valid_theoretical, 'mean_theoretical_power'] - 1)

    bin_means['perc_diff_fleet'] = np.nan
    bin_means.loc[valid_fleet, 'perc_diff_fleet'] = (bin_means.loc[valid_fleet, 'mean_real_power']
                                                     / bin_means.loc[valid_fleet, 'mean_fleet_power'] - 1)

    # average the bin-level relative differences within each period.
    bin_means['avg_perc_diff_theoretical'] = (bin_means.groupby('period', observed = True)['perc_diff_theoretical'].transform('mean'))
    bin_means['avg_perc_diff_fleet'] = (bin_means.groupby('period', observed = True)['perc_diff_fleet'].transform('mean'))

    # number of valid bins contributing to each period's average.
    bin_means['n_valid_theoretical_bins'] = (bin_means.groupby('period', observed = True)['perc_diff_theoretical'].transform('count'))
    bin_means['n_valid_fleet_bins'] = (bin_means.groupby('period', observed = True)['perc_diff_fleet'].transform('count'))

    return bin_means
```

```python
def scatter_power_curves(df, original, turbine, start = 0, ncols = 2, cutin = 4.0, rated = 15.0):
    t = original[(original['turbine'] == turbine) & (original['corrected_ws'] >= cutin) & (original['corrected_ws'] <= rated)]
    t['period'] = t['utc'].dt.to_period('M')
    
    periods = result['period'].unique()[start:start + 4]
    fig, axes = plt.subplots(2, 2, figsize = (12, 8), sharex = True, sharey = True)

    for ax, period in zip(axes.flat, periods):
        grp = df[df['period'] == period].sort_values('corrected_ws')
        grpt = t[t['period'] == period]

        ax.scatter(grpt['corrected_ws'], grpt['KPI_real_power_ac'], label = 'Real Power', color = 'cornflowerblue',s = 8, alpha = 0.6, zorder = 1)
        ax.plot(grp['corrected_ws'], grp['mean_theoretical_power'], label = 'Mean Theoretical Power', color = 'firebrick')
        ax.plot(grp['corrected_ws'], grp['mean_fleet_power'], label = 'Mean Fleet Power', color = 'black')
        ax.set_title(period.strftime('%B %Y'))
        ax.grid(alpha = 0.3)

    fig.supxlabel('Corrected Wind Speed')
    fig.supylabel('Mean Power Output')
    fig.suptitle(f'Mean Power Output by Binned Wind Speed: Turbine {turbine}')

    handles, labels = axes[0, 0].get_legend_handles_labels()
    fig.legend(handles, labels, loc = 'upper right')

    plt.tight_layout()
    plt.show()
```

```python
def pair_turbines(data, turbine, reference, low = None, up = None):
    # selects variables of interest
    cols = ['turbine', 'utc', 'wind_speed']
    # filters for turbine of interest and selects variables of interest
    a = data.loc[data['turbine'] == turbine, cols]
    # filters for reference and selects variables of interest
    b = data.loc[data['turbine'] == reference, cols]

    # inner merges the turbine with the reference on utc column
    out = a.merge(b, on = 'utc', how = 'inner', suffixes = ('_1', '_2'))
    out = out.rename(columns={'turbine_1': 'turbine', 'turbine_2': 'reference',
                              'wind_speed_1': 'wind_speed_turbine', 'wind_speed_2': 'wind_speed_reference'})

    # filters the wind speed if specified
    if low is not None:
        out = out[out['wind_speed_reference'] >= low]
    if up is not None:
        out = out[out['wind_speed_reference'] <= up]
    
    return out[['utc', 'turbine', 'reference', 'wind_speed_turbine', 'wind_speed_reference']]
```

```python
def group_metrics(data, group, reference, low = None, up = None):
    # filters for only the specified group and excludes the reference
    turbines = data.loc[(data['group'] == group) & (data['turbine'] != reference), 'turbine'].unique()

    # initializes results and applies pair_turbines() to each pair against reference
    results = []
    for turbine in turbines:
        p = pair_turbines(data, turbine, reference, low = low, up = up).dropna(subset = ['wind_speed_turbine', 'wind_speed_reference'])
        x, y = p['wind_speed_reference'], p['wind_speed_turbine'] 
        slope, intercept = np.polyfit(x, y, 1) if len(p) > 1 else (np.nan, np.nan)
        results.append({'turbine': turbine, 'n': len(p), 'bias': (y - x).mean(),
                        'corr': x.corr(y), 'slope': slope, 'intercept': intercept})
    
    return pd.DataFrame(results).sort_values('corr', ascending = False)
```

```python
def plot_pair(data, turbine, reference, month = None, low = None, up = None):
    # applies the pair_turbines() function and drops rows missing either wind speed
    p = pair_turbines(data, turbine, reference, low = low, up = up).dropna(subset = ['wind_speed_turbine', 'wind_speed_reference'])

    # filters to the specified month if given
    if month is not None:
        p = p.assign(utc = pd.to_datetime(p['utc']))
        m = pd.Period(month, freq = 'M')
        p = p[(p['utc'].dt.year == m.year) & (p['utc'].dt.month == m.month)]

    # selects the reference as x and the turbine as y
    x, y = p['wind_speed_reference'], p['wind_speed_turbine']

    # computes the slope and intercept between the turbine and reference
    slope, intercept = np.polyfit(x, y, 1)
    xs = np.linspace(x.min(), x.max(), 100)
    period = f', {month}' if month is not None else ''
    fig, ax = plt.subplots(figsize = (6, 6))
    ax.scatter(x, y, s = 5, alpha = 0.3)
    ax.plot(xs, slope * xs + intercept, color = 'red', label = f'y = {slope:.3f}x + {intercept:.3f}\nr = {x.corr(y):.3f}, n = {len(p)}')
    ax.plot(xs, xs, 'k--', lw = 1, label = '1:1 line')
    ax.set(xlabel = f'{reference} wind speed', ylabel = f'{turbine} wind speed', title = f'{turbine} vs {reference}{period}')
    ax.legend()
    plt.show()
```

```python
def plot_pair_ts(data, turbine, reference, month = None, low = None, up = None):
    # applies the pair_turbines() function to the inputs
    p = pair_turbines(data, turbine, reference, low = low, up = up).dropna(subset = ['wind_speed_turbine', 'wind_speed_reference'])
    p = p.sort_values('utc')

    # filters for the month if given
    if month is not None:
        m = pd.Period(month, freq = 'M')
        p = p[(p['utc'].dt.year == m.year) & (p['utc'].dt.month == m.month)]

    # defines the period and plots the time series
    period = f', {month}' if month is not None else ''
    fig, ax = plt.subplots(figsize = (12, 5))
    ax.plot(p['utc'], p['wind_speed_turbine'], lw = 1, label = f'{turbine}')
    ax.plot(p['utc'], p['wind_speed_reference'], lw = 1, alpha = 0.8, label = f'{reference}')
    ax.set(xlabel = 'Time (UTC)', ylabel = 'Wind speed', title = f'{turbine} vs {reference} wind speed{period}, n = {len(p)}')
    ax.legend()
    ax.grid(alpha = 0.3)
    fig.autofmt_xdate()
    plt.show()
```

```python
def plot_pair_diff_ts(data, turbine, reference, month = None, low = None, up = None, ma = '7D'):
    # applies the pair_turbines() function to the inputs
    p = pair_turbines(data, turbine, reference, low = low, up = up).dropna(subset = ['wind_speed_turbine', 'wind_speed_reference'])
    p = p.sort_values('utc')

    # filters for the month if given
    if month is not None:
        m = pd.Period(month, freq = 'M')
        p = p[(p['utc'].dt.year == m.year) & (p['utc'].dt.month == m.month)]

    # computes difference and plots
    p = p.assign(diff = p['wind_speed_turbine'] - p['wind_speed_reference'])
    period = f', {month}' if month is not None else ''
    fig, ax = plt.subplots(figsize = (12, 5))
    ax.plot(p['utc'], p['diff'], lw = 1, alpha = 0.7, label = f'Difference {turbine} - {reference}')

    # computes moving average of the time series process to see trends
    if ma is not None:
        rolling = p.set_index('utc')['diff'].rolling(ma).mean()
        ax.plot(rolling.index, rolling.values, color = 'red', lw = 1.5, label = f'{ma} moving average')
    ax.axhline(0, color = 'k', lw = 0.8)
    ax.set(xlabel = 'Time (UTC)', ylabel = 'Wind speed difference', title = f'{turbine} - {reference} wind speed difference{period}, n = {len(p)}')
    ax.legend()
    ax.grid(alpha = 0.3)
    fig.autofmt_xdate()
    plt.show()
```

```python
def summary_plots(data, turbine, reference, month = None, low = None, up = None, ma = '7D'):
    plot_pair(data, turbine, reference, month = month, low = low, up = up)
    plot_pair_diff_ts(data, turbine, reference, month = month, low = low, up = up, ma = ma)
    plt.show()
```
